# Taks 1: Speed Unit Conversion

```python
speed_kmh = 72
speed_ms = speed_kmh / 3.6
print(f"Speed: {speed_ms:.2f} m/s")
```

# Task 2: Power Unit Conversion

```python
power_kw = 12.5
power_w = power_kw * 1000
print(f"Power: {power_w:.2f} W")
```

# Task 3: Electrical Power and Energy Calculator

```python
voltage = 24.0
current = 8.5
time = 3.0

power = voltage * current
energy = power * time

print("===== ELECTRICAL CALCULATION =====")
print(f"Voltage     = {voltage:.2f} V")
print(f"Current     = {current:.2f} A")
print(f"Power       = {power:.2f} W")
print(f"Operating Time = {time:.2f} hour")
print(f"Energy      = {energy:.2f} Wh")
```

# Task 4: Comparison and Logical Operators

```python
voltage_value = 24.0
current_limit = 10.0

is_safe = current_limit <= 15.0
is_high_voltage = voltage_value > 50.0

print(f"Current is within safe limit (<= 15A): {is_safe}")
print(f"Voltage is high (> 50V): {is_high_voltage}")
```

# Task 5: Conditional Statements (if-elif-else)

```python
engine_load = 75.0  # dalam persen [%]

if engine_load < 50.0:
    status = "Low Load"
elif engine_load <= 80.0:
    status = "Normal Load"
else:
    status = "High Load / Overload"

print(f"Engine Load: {engine_load}%")
print(f"Operating Status: {status}")
```

# Task 6: Loop Structure (for / while)

```python
print("Monitoring Engine RPM for 5 seconds:")
for second in range(1, 6):
    rpm = 1000 + (second * 150)
    print(f"Time: {second} s -> Engine RPM: {rpm}")
```
