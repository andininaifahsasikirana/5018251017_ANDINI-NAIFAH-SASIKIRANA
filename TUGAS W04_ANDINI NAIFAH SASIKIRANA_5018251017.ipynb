# SENSOR LOG ANALYZER - SUHU MESIN KAPAL

```python

sensor_data = [32, 67, 84, 45, 73, -4, 58, 91, 39, 76,
               63, 87, 27, 71, 96, 54, -8, 69, 82, 42]

batas_alarm = 80

total = 0
jumlah_valid = 0
jumlah_alarm = 0
nilai_min = None
nilai_max = None

print("=== SENSOR LOG ANALYZER SUHU MESIN KAPAL ===")

for i in range(len(sensor_data)):
    suhu = sensor_data[i]

    if suhu < 0:
        print("Sampel", i + 1, "=", suhu, "°C | DATA TIDAK VALID")
        continue

    if suhu >= batas_alarm:
        status = "ALARM"
        jumlah_alarm += 1
    else:
        status = "NORMAL"

    print("Sampel", i + 1, "=", suhu, "°C |", status)

    total += suhu
    jumlah_valid += 1

    if nilai_min is None or suhu < nilai_min:
        nilai_min = suhu

    if nilai_max is None or suhu > nilai_max:
        nilai_max = suhu

print("\n=== HASIL ANALISIS ===")

if jumlah_valid > 0:
    rata_rata = total / jumlah_valid
    print("Jumlah data valid =", jumlah_valid)
    print("Rata-rata suhu mesin =", round(rata_rata, 2), "°C")
    print("Jumlah alarm =", jumlah_alarm)
    print("Suhu minimum =", nilai_min, "°C")
    print("Suhu maksimum =", nilai_max, "°C")
else:
    print("Tidak ada data suhu yang valid.")
```
